# Visualização de Sinais - Questões 18 e 19

Este notebook implementa o esboço dos sinais descritos nas questões 18 e 19 da lista de exercícios de Sinais e Sistemas.

```python
import numpy as np
import matplotlib.pyplot as plt

# Configurando o estilo dos gráficos para melhor visualização
plt.rcParams['figure.figsize'] = (16, 6)
plt.rcParams['axes.grid'] = True
```

## Questão 18
Para o sinal $x(t) = \delta(t + 2) - \delta(t - 2)$, o sinal $y(t) = \int_{-\infty}^{t} x(\tau) d\tau$ resulta em:

$$y(t) = u(t + 2) - u(t - 2)$$

Este é um pulso retangular que vale 1 no intervalo $t \in [-2, 2]$. A seguir, plotamos $y(t)$.

```python
# Definindo o vetor de tempo
t = np.linspace(-4, 4, 1000)

# Construindo y(t) = u(t+2) - u(t-2)
y = np.where((t >= -2) & (t <= 2), 1.0, 0.0)

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(t, y, 'b', linewidth=2, label='y(t) = u(t+2) - u(t-2)')
ax.fill_between(t, y, alpha=0.2, color='blue')
ax.set_title('Questão 18 - Sinal y(t)', fontsize=14)
ax.set_xlabel('t', fontsize=12)
ax.set_ylabel('y(t)', fontsize=12)
ax.set_ylim(-0.5, 1.5)
ax.axhline(0, color='black', linewidth=1)
ax.axvline(0, color='black', linewidth=1)
ax.legend()
plt.show()
```

## Questão 19
O sinal original $x(t)$ é definido por partes.
Vamos implementar a função $x(t)$ e em seguida plotar todas as transformações solicitadas lado a lado.

```python
def x(t):
    """
    Sinal original x(t) da Figura P1.21
    """
    y = np.zeros_like(t)
    y[(t >= -2) & (t < -1)] = -t[(t >= -2) & (t < -1)] - 2
    y[(t >= -1) & (t < 0)] = 1
    y[(t >= 0) & (t < 1)] = 2
    y[(t >= 1) & (t <= 2)] = -t[(t >= 1) & (t <= 2)] + 2
    return y

t19 = np.linspace(-6, 6, 2000)

# Plot do sinal original
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(t19, x(t19), 'k', linewidth=2, label='x(t) original')
ax.set_title('Sinal original x(t) - Figura P1.21', fontsize=14)
ax.set_xlabel('t', fontsize=12)
ax.set_ylabel('x(t)', fontsize=12)
ax.set_ylim(-1.5, 2.5)
ax.axhline(0, color='black', linewidth=1)
ax.axvline(0, color='black', linewidth=1)
ax.legend()
plt.show()
```

### Transformações de $x(t)$
Abaixo plotamos as letras **a** a **f** solicitadas na questão. Os gráficos são organizados lado a lado usando subplots.

```python
# Criando figura com subplots 2x3 para os 6 gráficos
fig, axs = plt.subplots(2, 3, figsize=(18, 10))
fig.suptitle('Questão 19 - Transformações do sinal x(t)', fontsize=16)

# Função auxiliar para padronizar o visual de cada eixo
def format_axis(ax, title):
    ax.set_title(title, fontsize=12, fontweight='bold')
    ax.set_xlabel('t')
    ax.set_ylim(-1.5, 3.5)  # Alterado para 3.5 devido à soma no item e)
    ax.axhline(0, color='black', linewidth=1.2)
    ax.axvline(0, color='black', linewidth=1.2)

# a) x(t - 1) -> Deslocamento de 1 unidade para a direita
y_a = x(t19 - 1)
axs[0, 0].plot(t19, y_a, 'r', linewidth=2)
format_axis(axs[0, 0], 'a) x(t - 1)')

# b) x(2 - t) = x(-(t - 2)) -> Inversão temporal e deslocamento de 2 para a direita
y_b = x(2 - t19)
axs[0, 1].plot(t19, y_b, 'g', linewidth=2)
format_axis(axs[0, 1], 'b) x(2 - t)')

# c) x(2t + 1) = x(2(t + 0.5)) -> Compressão por 2 e avanço de 0.5
y_c = x(2*t19 + 1)
axs[0, 2].plot(t19, y_c, 'b', linewidth=2)
format_axis(axs[0, 2], 'c) x(2t + 1)')

# d) x(4 - t/2) = x(-1/2 (t - 8)) -> Expansão por 2, inversão e deslocamento
y_d = x(4 - t19/2)
axs[1, 0].plot(t19, y_d, 'm', linewidth=2)
format_axis(axs[1, 0], 'd) x(4 - t/2)')

# e) [x(t) + x(-t)]u(t) -> Parte par * 2 para t>0 multiplicada pelo degrau
u_t = np.where(t19 >= 0, 1.0, 0.0)
y_e = (x(t19) + x(-t19)) * u_t
axs[1, 1].plot(t19, y_e, 'c', linewidth=2)
format_axis(axs[1, 1], 'e) [x(t) + x(-t)]u(t)')

# f) x(t)[\delta(t + 3/2) - \delta(t - 3/2)]
# Esse produto resulta em amostras de x(t) escalonando os impulsos em t=-1.5 e t=1.5
# Valores de x(t):
# x(-1.5) = -0.5
# x(1.5) = 0.5
# Resultando em: -0.5*\delta(t + 1.5) - 0.5*\delta(t - 1.5) (Notando o sinal do delta)
# Wait: x(1.5)*(-delta(t - 1.5)) -> -0.5 delta(t - 1.5)
# Assim plotamos impulsos (setas) nas posições usando stem
axs[1, 2].stem([-1.5, 1.5], [x(-1.5), -x(1.5)], linefmt='C8-', markerfmt='C8o', basefmt='k-')
format_axis(axs[1, 2], 'f) x(t)[\delta(t + 3/2) - \delta(t - 3/2)]')

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

## Questão 1.12 (Haykin)
Um sinal de pulso triangular $x(t)$ é descrito na figura P1.12. Vamos esboçar os sinais derivados de $x(t)$.

```python
def x_tri(t):
    """ Pulso triangular da figura P1.12 """
    return np.where(np.abs(t) <= 1, 1 - np.abs(t), 0.0)

t_tri = np.linspace(-4, 4, 1000)

fig, ax = plt.subplots(figsize=(6, 3))
ax.plot(t_tri, x_tri(t_tri), 'k', linewidth=2)
ax.set_title('Sinal original x(t) - Pulso triangular P1.12')
ax.set_xlabel('t')
ax.set_ylim(-0.5, 1.5)
ax.axhline(0, color='black')
ax.axvline(0, color='black')
plt.show()
```

### Transformações do Pulso Triangular

```python
fig, axs = plt.subplots(2, 3, figsize=(18, 10))
fig.suptitle('Questão 1.12 Haykin - Transformações', fontsize=16)

def format_ax_tri(ax, title):
    ax.set_title(title, fontweight='bold')
    ax.set_xlabel('t')
    ax.set_ylim(-0.5, 2.5)
    ax.axhline(0, color='black', linewidth=1)
    ax.axvline(0, color='black', linewidth=1)

# a) x(3t)
axs[0, 0].plot(t_tri, x_tri(3*t_tri), 'r', linewidth=2)
format_ax_tri(axs[0, 0], 'a) x(3t)')

# b) x(3t + 2)
axs[0, 1].plot(t_tri, x_tri(3*t_tri + 2), 'g', linewidth=2)
format_ax_tri(axs[0, 1], 'b) x(3t + 2)')

# c) x(-2t - 2)
axs[0, 2].plot(t_tri, x_tri(-2*t_tri - 2), 'b', linewidth=2)
format_ax_tri(axs[0, 2], 'c) x(-2t - 2)')

# d) x(2(t + 2))
axs[1, 0].plot(t_tri, x_tri(2*(t_tri + 2)), 'm', linewidth=2)
format_ax_tri(axs[1, 0], 'd) x(2(t + 2))')

# e) x(2(t - 2))
axs[1, 1].plot(t_tri, x_tri(2*(t_tri - 2)), 'c', linewidth=2)
format_ax_tri(axs[1, 1], 'e) x(2(t - 2))')

# f) x(3t) + x(3t + 2)
axs[1, 2].plot(t_tri, x_tri(3*t_tri) + x_tri(3*t_tri + 2), 'y', linewidth=2)
format_ax_tri(axs[1, 2], 'f) x(3t) + x(3t + 2)')

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

## Questão 1.22 (Oppenheim)
O sinal discreto $x[n]$ é mostrado na figura P1.22. Vamos defini-lo e plotar as transformações solicitadas.

```python
n = np.arange(-10, 11)

def x_disc(n_vals):
    y = np.zeros_like(n_vals, dtype=float)
    y[n_vals == -3] = -0.5
    y[n_vals == -2] = 0.5
    y[(n_vals >= -1) & (n_vals <= 2)] = 1.0
    y[n_vals == 3] = 0.5
    return y

fig, ax = plt.subplots(figsize=(6, 3))
ax.stem(n, x_disc(n), linefmt='k-', markerfmt='ko', basefmt='k-')
ax.set_title('Sinal original x[n] - Figura P1.22')
ax.set_xlabel('n')
ax.set_xlim(-5, 5)
ax.set_ylim(-1, 1.5)
plt.show()
```

### Transformações de Tempo Discreto

```python
fig, axs = plt.subplots(3, 3, figsize=(18, 15))
fig.suptitle('Questão 1.22 Oppenheim - Transformações Discretas', fontsize=16)

def format_ax_disc(ax, title):
    ax.set_title(title, fontweight='bold')
    ax.set_xlabel('n')
    ax.set_xlim(-8, 8)
    ax.set_ylim(-1, 2)
    ax.axhline(0, color='black', linewidth=1)
    ax.axvline(0, color='black', linewidth=1)

# a) x[n - 4]
axs[0, 0].stem(n, x_disc(n - 4), linefmt='C0-', markerfmt='C0o', basefmt='k-')
format_ax_disc(axs[0, 0], 'a) x[n - 4]')

# b) x[3 - n]
axs[0, 1].stem(n, x_disc(3 - n), linefmt='C1-', markerfmt='C1o', basefmt='k-')
format_ax_disc(axs[0, 1], 'b) x[3 - n]')

# c) x[3n]
axs[0, 2].stem(n, x_disc(3 * n), linefmt='C2-', markerfmt='C2o', basefmt='k-')
format_ax_disc(axs[0, 2], 'c) x[3n]')

# d) x[3n + 1]
axs[1, 0].stem(n, x_disc(3 * n + 1), linefmt='C3-', markerfmt='C3o', basefmt='k-')
format_ax_disc(axs[1, 0], 'd) x[3n + 1]')

# e) x[n]u[3 - n]
u_3_n = np.where(3 - n >= 0, 1.0, 0.0)
axs[1, 1].stem(n, x_disc(n) * u_3_n, linefmt='C4-', markerfmt='C4o', basefmt='k-')
format_ax_disc(axs[1, 1], 'e) x[n]u[3 - n]')

# f) x[n - 2] \delta[n - 2]
delta_n_2 = np.where(n == 2, 1.0, 0.0)
axs[1, 2].stem(n, x_disc(n - 2) * delta_n_2, linefmt='C5-', markerfmt='C5o', basefmt='k-')
format_ax_disc(axs[1, 2], 'f) x[n - 2]\delta[n - 2]')

# g) 1/2 x[n] + 1/2 (-1)^n x[n]
y_g = 0.5 * x_disc(n) + 0.5 * ((-1)**n) * x_disc(n)
axs[2, 0].stem(n, y_g, linefmt='C6-', markerfmt='C6o', basefmt='k-')
format_ax_disc(axs[2, 0], 'g) 1/2 x[n] + 1/2 (-1)^n x[n]')

# h) x[(n - 1)^2]
y_h = x_disc((n - 1)**2)
axs[2, 1].stem(n, y_h, linefmt='C7-', markerfmt='C7o', basefmt='k-')
format_ax_disc(axs[2, 1], 'h) x[(n - 1)^2]')

# Remove empty 9th subplot
fig.delaxes(axs[2, 2])

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```
