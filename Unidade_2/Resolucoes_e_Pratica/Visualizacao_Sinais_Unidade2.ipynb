# Visualização Gráfica — 2ª Lista de Exercícios (Unidade 2)
Este notebook implementa o esboço de todas as convoluções em tempo discreto e contínuo da 2ª Lista de Exercícios de TEC 412 (Sinais e Sistemas).


```python
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams['figure.figsize'] = (10, 4)
plt.rcParams['axes.grid'] = True
```

## Questão 1.1 (Exercício 1.1 em Sala de Aula — Apostila, pág. 14)
Convolução discreta [n] = x[n] * h[n]$ para [n] = -u[n] + 2u[n-3] - u[n-6]$ e [n] = u[n+1] - u[n-10]$.


```python
n = np.arange(-4, 19)
y = np.zeros_like(n, dtype=float)
for i, val in enumerate(n):
    if -1 <= val <= 1: y[i] = -val - 2
    elif 2 <= val <= 4: y[i] = val - 4
    elif 5 <= val <= 8: y[i] = 0
    elif 9 <= val <= 11: y[i] = val - 9
    elif 12 <= val <= 14: y[i] = 15 - val
plt.figure(figsize=(10, 4))
plt.stem(n, y, basefmt='k-')
plt.title('Questão 1.1: Saída y[n] = x[n] * h[n]')
plt.xlabel('n'); plt.ylabel('y[n]'); plt.xticks(n)
plt.show()
```

## Questão 2.1 — [Oppenheim 2.3, pág. 138]
[n] = 2[1 - (1/2)^{n+1}]u[n]$


```python
n = np.arange(-4, 12)
y = np.where(n >= 0, 2.0 * (1.0 - (0.5)**(n + 1.0)), 0.0)
plt.figure(figsize=(10, 4))
plt.stem(n, y, basefmt='k-')
plt.axhline(2.0, color='r', linestyle='--', label='Assíntota = 2')
plt.title('Oppenheim 2.3: y[n] = 2[1 - (1/2)^(n+1)]u[n]')
plt.xlabel('n'); plt.ylabel('y[n]'); plt.xticks(n); plt.legend()
plt.show()
```

## Questão 2.2 — [Oppenheim 2.4, pág. 138]
Convolução de dois pulsos retangulares discretos ( \le n \le 23$).


```python
n = np.arange(4, 27)
y = np.zeros_like(n, dtype=float)
for i, val in enumerate(n):
    if 7 <= val <= 11: y[i] = val - 6
    elif 12 <= val <= 18: y[i] = 6
    elif 19 <= val <= 23: y[i] = 24 - val
plt.figure(figsize=(10, 4))
plt.stem(n, y, basefmt='k-')
plt.title('Oppenheim 2.4: y[n] = x[n] * h[n]')
plt.xlabel('n'); plt.ylabel('y[n]'); plt.xticks(n)
plt.show()
```

## Questão 2.3 — [Oppenheim 2.6, pág. 138]
[n] = 3^n/2$ para  < 0$ e /2$ para  \ge 0$.


```python
n = np.arange(-6, 8)
y = np.where(n < 0, (3.0**n)/2.0, 0.5)
plt.figure(figsize=(10, 4))
plt.stem(n, y, basefmt='k-')
plt.title('Oppenheim 2.6: y[n] = x[n] * h[n]')
plt.xlabel('n'); plt.ylabel('y[n]'); plt.xticks(n)
plt.show()
```

## Questão 1.2 (Exercício 1.2 em Sala de Aula — Apostila, pág. 23)
Convolução contínua (t) = u(t-1) - u(t-4)$ com (t) = u(t) + u(t-1) - 2u(t-2)$.


```python
t = np.linspace(-1, 8, 1000)
y = np.piecewise(t, [t < 1, (t >= 1) & (t < 2), (t >= 2) & (t < 3), (t >= 3) & (t < 4), (t >= 4) & (t < 5), (t >= 5) & (t < 6), t >= 6],
                 [0, lambda t: t - 1, lambda t: 2*t - 3, 3, lambda t: 7 - t, lambda t: 12 - 2*t, 0])
plt.figure(figsize=(10, 4))
plt.plot(t, y, 'b-', linewidth=2.2)
plt.fill_between(t, y, alpha=0.15, color='b')
plt.title('Questão 1.2: y(t) = x(t) * h(t)')
plt.xlabel('t (s)'); plt.ylabel('y(t)'); plt.xticks(range(-1, 9))
plt.show()
```

## Questão 3.1 — [Oppenheim 2.22 (a, b, c, d, e)] e Questão 5.1 [Haykin 2.11(b)]
Esboços completos das cinco convoluções em tempo contínuo da questão 2.22 e da resposta global de Haykin 2.11(b).


```python
fig, axes = plt.subplots(3, 2, figsize=(14, 11))
# (a)
t = np.linspace(-1, 6, 800)
y_a1 = np.where(t >= 0, (np.exp(-1.0*t) - np.exp(-2.0*t))/(2.0 - 1.0), 0.0)
y_a2 = np.where(t >= 0, t * np.exp(-1.0*t), 0.0)
axes[0,0].plot(t, y_a1, 'b-', lw=2, label=r'alpha=1, beta=2')
axes[0,0].plot(t, y_a2, 'r--', lw=2, label=r'alpha=beta=1')
axes[0,0].set_title('2.22(a): Exponenciais causais'); axes[0,0].legend()
# (b)
t_b = np.linspace(-2, 7, 1000)
y_b = np.piecewise(t_b, [t_b < 1, (t_b >= 1) & (t_b < 3), (t_b >= 3) & (t_b < 6), t_b >= 6],
                   [lambda t: 0.5*(np.exp(2*t) - 2*np.exp(2*(t-2)) + np.exp(2*(t-5))),
                    lambda t: 0.5*(np.exp(2) - 2*np.exp(2*(t-2)) + np.exp(2*(t-5))),
                    lambda t: 0.5*(np.exp(2*(t-5)) - np.exp(2)), 0.0])
axes[0,1].plot(t_b, y_b, 'g-', lw=2)
axes[0,1].set_title('2.22(b): x(t) em escada * e^(2t)u(1-t)')
# (c)
t_c = np.linspace(0, 6, 1000)
y_c = np.piecewise(t_c, [t_c < 1, (t_c >= 1) & (t_c < 3), (t_c >= 3) & (t_c < 5), t_c >= 5],
                   [0.0, lambda t: (2/np.pi)*(1 + np.cos(np.pi*t)), lambda t: -(2/np.pi)*(1 + np.cos(np.pi*t)), 0.0])
axes[1,0].plot(t_c, y_c, 'm-', lw=2)
axes[1,0].set_title('2.22(c): Ciclo senoidal * pulso retangular')
# (d)
t_d = np.linspace(-3, 4, 500)
axes[1,1].plot(t_d, 1.5*t_d + 1.0, 'k-', lw=2, label='y(t) = x(t) = at + b (a=1.5, b=1)')
axes[1,1].set_title('2.22(d): Rampa linear reproduzida exatamente'); axes[1,1].legend()
# (e)
t_e = np.linspace(-2.5, 3.5, 1200)
t_mod = ((t_e + 0.5) % 2.0) - 0.5
y_e = np.where(t_mod < 0.5, 0.5 - (t_mod - 0.5)**2, (t_mod - 1.5)**2 - 0.5)
axes[2,0].plot(t_e, y_e, 'c-', lw=2.2)
axes[2,0].set_title('2.22(e): Onda quadrada periódica * pulso triangular')
# Haykin 2.11(b)
n_h = np.arange(-4, 7)
h_eq = np.where(n_h < -2, 0.0, np.where(n_h <= 2, 8.0 - (0.5)**n_h, 31.0/4.0))
axes[2,1].stem(n_h, h_eq, basefmt='k-')
axes[2,1].set_title('Haykin 2.11(b): Resposta global h[n]')
plt.tight_layout()
plt.show()
```
